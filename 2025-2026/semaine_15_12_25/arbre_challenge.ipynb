### Challenge 1 : Organiser des données client

```python
clients = [
    # CORRECTION 1 : Suppression des espaces dans les clés ("id", pas " id ")
    { "id": 1, "age": 25, "ville": "Paris", "departement": "75" },
    { "id": 2, "age": 35, "ville": "Lyon", "departement": "69" },
    { "id": 3, "age": 28, "ville": "Paris", "departement": "75" },
    { "id": 4, "age": 42, "ville": "Marseille", "departement": "13" },
    { "id": 5, "age": 31, "ville": "Lyon", "departement": "69" }
]

arbre = {"France": {}}

for client in clients:
    dept_key = f"{client['departement']} ({client['ville']})"

    if client["age"] < 30:
        groupe_age = "<30 ans"
    elif 30 <= client["age"] <= 40:
        groupe_age = "30-40 ans"
    else:
        groupe_age = ">40 ans"

    if dept_key not in arbre["France"]:
        arbre["France"][dept_key] = {}

    if groupe_age not in arbre["France"][dept_key]:
        arbre["France"][dept_key][groupe_age] = []

    arbre["France"][dept_key][groupe_age].append(f"Client {client['id']}")

print("--- STRUCTURE ET COMPTAGE ---")
total_clients = 0

for pays, depts in arbre.items():
    print(pays)

    for dept_nom, ages in depts.items():
        nb_clients_dept = sum(len(c) for c in ages.values())
        print(f"  └── {dept_nom} : {nb_clients_dept} clients")

        for age_nom, liste_clients in ages.items():
            nb = len(liste_clients)
            total_clients += nb
            print(f"      └── {age_nom} : {nb}")
            print(f"          └── {', '.join(liste_clients)}")

print(f"\nTotal général France : {total_clients} clients")
```

### Challenge 2 : Arbre de décision simple

```python
data = [
    { "age": 25, "ville": "Paris", "achat": "oui" },
    { "age": 35, "ville": "Lyon", "achat": "oui" },
    { "age": 45, "ville": "Paris", "achat": "non" },
    { "age": 28, "ville": "Marseille", "achat": "non" },
    { "age": 31, "ville": "Lyon", "achat": "oui" },
    { "age": 40, "ville": "Paris", "achat": "non" }
]

def arbre_decision(client):
    """
    Fonction qui représente l'arbre de décision manuel.
    Retourne 'oui' ou 'non'.
    """
    if client["ville"] == "Lyon":
        return "oui"
    
    elif client["age"] < 30:
        return "oui"
    else:
        return "non"

# --- CORRECTION ICI : DÉFINITION DE LA LISTE MANQUANTE ---
nouveaux_clients = [
    { "age": 27, "ville": "Paris" },
    { "age": 38, "ville": "Marseille" }
]
# ---------------------------------------------------------

print("--- PRÉDICTIONS ---")
for client in nouveaux_clients:
    prediction = arbre_decision(client)
    print(f"Client ({client['age']} ans, {client['ville']}) -> Achat prédit : {prediction}")


print("\n--- CALCUL DE L'ACCURACY ---")
bonnes_predictions = 0

for row in data:
    prediction = arbre_decision(row)
    realite = row["achat"]
    
    if prediction == realite:
        bonnes_predictions += 1
        resultat = "✅ Correct"
    else:
        resultat = "❌ Erreur"
        
    print(f"Client {row['age']} ans/{row['ville']} : Prédit '{prediction}' vs Réel '{realite}' -> {resultat}")

accuracy = (bonnes_predictions / len(data)) * 100
print(f"\nAccuracy du modèle : {accuracy:.2f}%")
```

### Challenge 4 : Segmentation clients avec arbre

```python
ventes = [
    { "client": "Alice",   "montant": 150, "frequence": 3, "segment": "A" },
    { "client": "Bob",     "montant": 80,  "frequence": 5, "segment": "B" },
    { "client": "Charlie", "montant": 200, "frequence": 2, "segment": "A" },
    { "client": "Diana",   "montant": 50,  "frequence": 1, "segment": "C" },
    { "client": "Eve",     "montant": 120, "frequence": 4, "segment": "B" }
]

def segmenter_client(montant, frequence):
    """
    Arbre de décision pour la segmentation.
    Retourne le segment (A, B ou C).
    """
    if montant > 100 and frequence > 2:
        return "A"
    
    elif montant > 100 or frequence > 2:
        return "B"
    
    else:
        return "C"

nouveaux_clients = [
    {"montant": 90,  "frequence": 4},
    {"montant": 110, "frequence": 1},
    {"montant": 60,  "frequence": 2}
]

print("--- CLASSEMENT DES NOUVEAUX CLIENTS ---")
for i, cl in enumerate(nouveaux_clients):
    seg = segmenter_client(cl["montant"], cl["frequence"])
    print(f"Client {i+1} (Montant: {cl['montant']}, Freq: {cl['frequence']}) -> Segment {seg}")


print("\n--- RÉPARTITION (Basée sur nos règles) ---")
repartition = {"A": 0, "B": 0, "C": 0}

for vente in ventes:
    seg_calcule = segmenter_client(vente["montant"], vente["frequence"])
    repartition[seg_calcule] += 1

for seg, count in repartition.items():
    print(f"Segment {seg} : {count} clients")
```
