# Enoncé

Vous avez un jeu de donnée d'essais moteurs. Chaque jour, plusieurs capteurs mesurent des informations dans différentes parties du moteur.
Le métier souhaite avoir un moyen automatique de détecter des anomalie (indicateur statistique ou algorithme de machine learning).

Vous pouvez trouver dans "data" les données des capteurs pour différentes journées.
Vous disposez en plus des informations suivantes :

    - les experiences ne se ressemblent pas : les experiences de demain pourront être très différentes des experiences passées
    - les experts métiers savent déterminer un capteur anormal en comparant ses caractéristiques (moyennes, fréquences, bruits) à d'autres capteurs normalement proches
    - dans les données fournie, le nom de la colonne est au format "{nomcapteur}_{status}", avec status valant "NORMAL" ou "ANORMAL".


A partir de ces informations, mettez au point une méthode pour déterminer les anomalies basée sur les jours 0 à 7:

    - chargez les données, et visualisez les (par exemple, visualisez des séries temporelles du même capteur, de plusieurs capteurs, normales vs anormales, ...).
    - trouver et faite une fonction permettant de trouver les capteurs proches d'un capteur donné
    - déterminez un indicateur de similarité d'un capteur avec ses voisins
    - tester le pourcentage d'anomalie que votre algorithme/indicateur permet de détecter sur les jours 8 et 9, puis améliorez votre algorithme/indicateur.
