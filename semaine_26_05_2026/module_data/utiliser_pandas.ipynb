#### Les manipulations à connaitre

- Selectionner les lignes respectant une condition

```python
df[df["Age"] > 90]
```

- respectant plusieurs conditions

```python
df[(df["Age"] > 90) & (df["Year"]<1930)]  # et = &, ou = |
```

- Compter le nombre d'occurences de chacunes des valeurs possibles d'une colonne

```python
df.Medal.value_counts()
```

- Remplacer les valeurs manquantes (egales "Nan") par une autre valeur

```python
df.Medal = df.Medal.fillna(value=0)
```

- Autre solution : supprimer les lignes avec des valeurs manquantes

```python
df = df.dropna(subset=['Medal'])
```

- Affichez la valeur la plus fréquente d'une colonne (par exemple le sportif le plus présent du jeu de donnée)

```python
best_sportif = df.Name.mode()
print(best_sportif[0])
```

- Affichez valeurs d'une colonne pour les lignes respectant une condition sur une autre colonne. Exemple : les sports du meilleur sportif.

```python
df['Sport'][df['Name']==best_sportif[0]]
```

- le min et max d'une colonne

```python
df['Age'].max()
```

- La ligne du sportif le plus petit

```python
line_smallest = df['Height'].idxmin()
df.loc[line_smallest]
```

- La moyenne, medianne, ...

```python
df.Age.mean()
```

- Grouper sur une colonne, et appliquer un indicateur sur chaque groupe. Ex: l'age min et max par sport

```python
group = df.groupby('Sport')
print(group)
```

```python
df_agg = group.Age.agg(['min', 'max'])
df_agg
```

- Faire des opérations entre colonnes. Exemple : soustraire une colonne à une autre

```python
df_agg["diff"] = df_agg["max"] - df_agg["min"]
df_agg
```

### Fusionner plusieurs dataframe 

```python
# import csv 
df_continent = pd.read_csv("pays_continents.csv", encoding = "ISO-8859-1",  sep=";" )
```

Merge permet de fusionner les deux dataframes en fournissant le nom de la colonne suivant laquelle faire la fusion

```python
df_merge = df.merge(df_continent, left_on="City", right_on="Ville", how='left')
df_merge
```
